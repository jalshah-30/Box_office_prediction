```python
df = df.sort_values('release_date').reset_index(drop=True)

df['director_avg_boxoffice'] = (
    df.groupby('directors')['collection']
      .transform(lambda x: x.expanding().mean().shift(1))
)

df['director_prior_movie_count'] = (
    df.groupby('directors').cumcount()
)
```

```python
df.head()
```

```python
df['lead_actor'] = df['cast'].apply(lambda x: x.split(',')[0].strip())

df['actor_avg_boxoffice'] = (
    df.groupby('lead_actor')['collection']
      .transform(lambda x: x.expanding().mean().shift(1))
)
df['actor_prior_movie_count'] = df.groupby('lead_actor').cumcount()
```

```python
global_expanding_mean = df['collection'].expanding().mean().shift(1)
df['director_avg_boxoffice'] = df['director_avg_boxoffice'].fillna(global_expanding_mean)
df['actor_avg_boxoffice'] = df['actor_avg_boxoffice'].fillna(global_expanding_mean)

df['director_is_debut'] = df['director_prior_movie_count'] == 0
df['actor_is_debut'] = df['actor_prior_movie_count'] == 0
```
